```python
from pyspark.sql import functions as F

rows = 5000

df_5000 = (
    spark.range(1, rows + 1)
    .withColumnRenamed("id", "employee_id")
    .withColumn("first_name",
                F.expr("""
                CASE floor(rand()*10)
                    WHEN 0 THEN 'John'
                    WHEN 1 THEN 'Amy'
                    WHEN 2 THEN 'David'
                    WHEN 3 THEN 'Sarah'
                    WHEN 4 THEN 'Mike'
                    WHEN 5 THEN 'Priya'
                    WHEN 6 THEN 'Ravi'
                    WHEN 7 THEN 'Neha'
                    WHEN 8 THEN 'Amit'
                    ELSE 'Anjali'
                END
                """))
    .withColumn("last_name",
                F.expr("""
                CASE floor(rand()*10)
                    WHEN 0 THEN 'Smith'
                    WHEN 1 THEN 'Lewis'
                    WHEN 2 THEN 'Brown'
                    WHEN 3 THEN 'Patel'
                    WHEN 4 THEN 'Sharma'
                    WHEN 5 THEN 'Kumar'
                    WHEN 6 THEN 'Singh'
                    WHEN 7 THEN 'Gupta'
                    WHEN 8 THEN 'Jain'
                    ELSE 'Mehta'
                END
                """))
    .withColumn("department",
                F.expr("""
                CASE floor(rand()*5)
                    WHEN 0 THEN 'IT'
                    WHEN 1 THEN 'Finance'
                    WHEN 2 THEN 'Marketing'
                    WHEN 3 THEN 'HR'
                    ELSE 'Sales'
                END
                """))
    .withColumn("city",
                F.expr("""
                CASE floor(rand()*5)
                    WHEN 0 THEN 'Mumbai'
                    WHEN 1 THEN 'Pune'
                    WHEN 2 THEN 'Bangalore'
                    WHEN 3 THEN 'Delhi'
                    ELSE 'Hyderabad'
                END
                """))
    .withColumn("country", F.lit("India"))
    .withColumn("salary", (F.rand()*80000 + 30000).cast("int"))
    .withColumn("experience", (F.rand()*20).cast("int"))
    .withColumn("joining_date",
                F.date_sub(F.current_date(),
                           (F.rand()*3000).cast("int")))
    .withColumn("email",
                F.concat(F.lit("user"),
                         F.col("employee_id"),
                         F.lit("@company.com")))
    .withColumn("manager_id",
                (F.rand()*200).cast("int"))
    .withColumn("status",
                F.when(F.rand() > 0.2, "Active")
                 .otherwise("Inactive"))
    .withColumn("processDate",
                F.current_timestamp())
)

df_5000.write.mode("overwrite") \
    .option("header", "true") \
    .csv("/Volumes/pp_learning/default/data_csv/employee4")
```
